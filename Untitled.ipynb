```python
"""
Wave Model — Character-level language model on Tiny Shakespeare
Architecture:
  - Fourier encoding of characters (sine/cosine components)
  - Position encoded as phase shift (native, not bolted on)
  - Frequency filter hidden layer
  - Output projection to vocab

Usage:
    pip install numpy requests
    python wave_model.py

Results are printed to stdout and saved to wave_results.json
"""

import numpy as np
import json
import time
import urllib.request
import os

# ─────────────────────────────────────────────
# CONFIG
# ─────────────────────────────────────────────
CTX_LEN      = 16       # context window (characters)
N_FREQS      = 16       # Fourier frequency components per character
HIDDEN       = 128      # frequency filter hidden units
EPOCHS       = 40
BATCH_SIZE   = 64
LR           = 0.001
EVAL_EVERY   = 1        # epochs between evals
TRAIN_SPLIT  = 0.9
SEED         = 42
MAX_CHARS    = 200_000  # use first N chars of dataset (speed vs thoroughness)

np.random.seed(SEED)

# ─────────────────────────────────────────────
# DATA
# ─────────────────────────────────────────────
DATA_URL  = "https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt"
DATA_PATH = "tinyshakespeare.txt"

def get_data():
    if not os.path.exists(DATA_PATH):
        print("Downloading Tiny Shakespeare...")
        urllib.request.urlretrieve(DATA_URL, DATA_PATH)
        print("Done.")
    with open(DATA_PATH, "r") as f:
        return f.read()

text = get_data()[:MAX_CHARS]
chars = sorted(set(text))
VOCAB_SIZE = len(chars)
c2i = {c: i for i, c in enumerate(chars)}
i2c = {i: c for c, i in c2i.items()}
data = np.array([c2i[c] for c in text], dtype=np.int32)

split = int(len(data) * TRAIN_SPLIT)
train_data = data[:split]
val_data   = data[split:]

print(f"Vocab size: {VOCAB_SIZE}")
print(f"Train chars: {len(train_data):,}  Val chars: {len(val_data):,}")

def get_batch(arr, batch_size, ctx_len):
    ix = np.random.randint(0, len(arr) - ctx_len - 1, size=batch_size)
    x  = np.stack([arr[i:i+ctx_len]   for i in ix])
    y  = np.array([arr[i+ctx_len]      for i in ix])
    return x, y

# ─────────────────────────────────────────────
# WAVE ENCODING
# ─────────────────────────────────────────────
INPUT_DIM = CTX_LEN * N_FREQS * 2   # sin+cos per freq, per position

def encode_batch(x_batch):
    """
    x_batch: (B, CTX_LEN) integer indices
    Returns: (B, INPUT_DIM) float32 Fourier features
    """
    B, T = x_batch.shape
    freqs = np.arange(1, N_FREQS + 1, dtype=np.float32)  # (N_FREQS,)
    
    # Character frequencies scaled to vocab
    char_angles = (2 * np.pi * freqs[None, None, :] * 
                   x_batch[:, :, None].astype(np.float32) / VOCAB_SIZE)  # (B, T, N_FREQS)
    
    # Phase shift from position — position is native to the wave
    pos = np.arange(T, dtype=np.float32)
    phase = (pos[None, :, None] * np.pi / T)  # (1, T, 1)
    
    angles = char_angles + phase  # (B, T, N_FREQS)
    
    sin_feats = np.sin(angles)  # (B, T, N_FREQS)
    cos_feats = np.cos(angles)  # (B, T, N_FREQS)
    
    # Interleave sin/cos, then flatten T and freq dims
    features = np.concatenate([sin_feats, cos_feats], axis=-1)  # (B, T, 2*N_FREQS)
    return features.reshape(B, -1).astype(np.float32)           # (B, INPUT_DIM)

# ─────────────────────────────────────────────
# MODEL PARAMETERS
# ─────────────────────────────────────────────
def init_params():
    scale1 = np.sqrt(2.0 / INPUT_DIM)
    scale2 = np.sqrt(2.0 / HIDDEN)
    return {
        # Frequency filter layer
        "W1": np.random.randn(INPUT_DIM, HIDDEN).astype(np.float32) * scale1,
        "b1": np.zeros(HIDDEN, dtype=np.float32),
        # Output projection
        "W2": np.random.randn(HIDDEN, VOCAB_SIZE).astype(np.float32) * scale2,
        "b2": np.zeros(VOCAB_SIZE, dtype=np.float32),
    }

def count_params(p):
    return sum(v.size for v in p.values())

params = init_params()
print(f"Wave model parameters: {count_params(params):,}")

# ─────────────────────────────────────────────
# FORWARD / BACKWARD
# ─────────────────────────────────────────────
def softmax(x):
    x = x - x.max(axis=-1, keepdims=True)
    e = np.exp(x)
    return e / e.sum(axis=-1, keepdims=True)

def forward(x_batch, p):
    x   = encode_batch(x_batch)           # (B, INPUT_DIM)
    h   = x @ p["W1"] + p["b1"]          # (B, HIDDEN)
    h   = np.maximum(0, h)                # ReLU — frequency gating
    logits = h @ p["W2"] + p["b2"]       # (B, VOCAB_SIZE)
    return logits, h, x

def cross_entropy_loss(logits, targets):
    probs = softmax(logits)
    B = len(targets)
    loss = -np.log(probs[np.arange(B), targets] + 1e-10).mean()
    return loss, probs

def backward(probs, targets, h, x, p, lr):
    B = len(targets)
    # Output layer gradient
    dlogits = probs.copy()
    dlogits[np.arange(B), targets] -= 1
    dlogits /= B

    # W2, b2
    p["W2"] -= lr * (h.T @ dlogits)
    p["b2"] -= lr * dlogits.sum(axis=0)

    # Hidden layer gradient
    dh = dlogits @ p["W2"].T
    dh *= (h > 0)  # ReLU grad

    # W1, b1
    p["W1"] -= lr * (x.T @ dh)
    p["b1"] -= lr * dh.sum(axis=0)

# ─────────────────────────────────────────────
# EVALUATION
# ─────────────────────────────────────────────
def evaluate(arr, p, n_batches=50):
    total_loss = 0
    correct = 0
    total = 0
    for _ in range(n_batches):
        xb, yb = get_batch(arr, BATCH_SIZE, CTX_LEN)
        logits, _, _ = forward(xb, p)
        loss, probs = cross_entropy_loss(logits, yb)
        total_loss += loss
        correct += (logits.argmax(axis=-1) == yb).sum()
        total += len(yb)
    return total_loss / n_batches, correct / total

# ─────────────────────────────────────────────
# TRAINING LOOP
# ─────────────────────────────────────────────
history = []
steps_per_epoch = len(train_data) // (BATCH_SIZE * CTX_LEN)

print(f"\n{'─'*55}")
print(f"{'Epoch':<8} {'Train Loss':<14} {'Val Loss':<14} {'Val Acc':<10} {'Time'}")
print(f"{'─'*55}")

train_start = time.time()

for epoch in range(1, EPOCHS + 1):
    ep_start = time.time()
    train_loss_acc = 0

    for step in range(steps_per_epoch):
        xb, yb = get_batch(train_data, BATCH_SIZE, CTX_LEN)
        logits, h, x = forward(xb, params)
        loss, probs = cross_entropy_loss(logits, yb)
        backward(probs, yb, h, x, params, LR)
        train_loss_acc += loss

    train_loss = train_loss_acc / steps_per_epoch

    if epoch % EVAL_EVERY == 0:
        val_loss, val_acc = evaluate(val_data, params)
        ep_time = time.time() - ep_start
        print(f"{epoch:<8} {train_loss:<14.4f} {val_loss:<14.4f} {val_acc*100:<10.2f} {ep_time:.1f}s")
        history.append({
            "epoch": epoch,
            "train_loss": float(train_loss),
            "val_loss": float(val_loss),
            "val_acc": float(val_acc),
        })

total_time = time.time() - train_start
print(f"{'─'*55}")
print(f"Total training time: {total_time:.1f}s")

# ─────────────────────────────────────────────
# SAMPLE GENERATION
# ─────────────────────────────────────────────
def generate(p, seed_text, n_chars=200, temperature=0.8):
    context = [c2i.get(c, 0) for c in seed_text[-CTX_LEN:]]
    context = ([0] * (CTX_LEN - len(context)) + context)[-CTX_LEN:]
    out = seed_text

    for _ in range(n_chars):
        xb = np.array([context])
        logits, _, _ = forward(xb, p)
        logits = logits[0] / temperature
        probs = softmax(logits)
        next_idx = np.random.choice(VOCAB_SIZE, p=probs)
        context = context[1:] + [next_idx]
        out += i2c[next_idx]

    return out

print("\n─── SAMPLE OUTPUT (temperature=0.8) ───")
sample = generate(params, "HAMLET:\n", n_chars=300)
print(sample)

# ─────────────────────────────────────────────
# SAVE RESULTS
# ─────────────────────────────────────────────
results = {
    "model": "wave",
    "parameters": count_params(params),
    "ctx_len": CTX_LEN,
    "n_freqs": N_FREQS,
    "hidden": HIDDEN,
    "vocab_size": VOCAB_SIZE,
    "epochs": EPOCHS,
    "total_time_s": total_time,
    "final_val_loss": history[-1]["val_loss"] if history else None,
    "final_val_acc": history[-1]["val_acc"] if history else None,
    "history": history,
}

with open("wave_results.json", "w") as f:
    json.dump(results, f, indent=2)

print(f"\nResults saved to wave_results.json")
print(f"Parameters: {count_params(params):,}")
print(f"Final val loss: {history[-1]['val_loss']:.4f}")
print(f"Final val acc:  {history[-1]['val_acc']*100:.2f}%")
```

```python
"""
Transformer — Character-level language model on Tiny Shakespeare
Architecture:
  - Token embedding table
  - Sinusoidal positional encoding (added on top)
  - Single attention head
  - Feedforward layer
  - Output projection

Matched to wave model in training setup for fair comparison.

Usage:
    pip install numpy requests
    python transformer_model.py

Results saved to transformer_results.json
"""

import numpy as np
import json
import time
import urllib.request
import os

# ─────────────────────────────────────────────
# CONFIG — matched to wave_model.py
# ─────────────────────────────────────────────
CTX_LEN      = 16
D_MODEL      = 64       # embedding dimension
D_FF         = 128      # feedforward hidden size
N_HEADS      = 2        # attention heads
EPOCHS       = 40
BATCH_SIZE   = 64
LR           = 0.001
EVAL_EVERY   = 1
TRAIN_SPLIT  = 0.9
SEED         = 42
MAX_CHARS    = 200_000

np.random.seed(SEED)

# ─────────────────────────────────────────────
# DATA — identical to wave script
# ─────────────────────────────────────────────
DATA_URL  = "https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt"
DATA_PATH = "tinyshakespeare.txt"

def get_data():
    if not os.path.exists(DATA_PATH):
        print("Downloading Tiny Shakespeare...")
        urllib.request.urlretrieve(DATA_URL, DATA_PATH)
        print("Done.")
    with open(DATA_PATH, "r") as f:
        return f.read()

text = get_data()[:MAX_CHARS]
chars = sorted(set(text))
VOCAB_SIZE = len(chars)
c2i = {c: i for i, c in enumerate(chars)}
i2c = {i: c for c, i in c2i.items()}
data = np.array([c2i[c] for c in text], dtype=np.int32)

split = int(len(data) * TRAIN_SPLIT)
train_data = data[:split]
val_data   = data[split:]

print(f"Vocab size: {VOCAB_SIZE}")
print(f"Train chars: {len(train_data):,}  Val chars: {len(val_data):,}")

def get_batch(arr, batch_size, ctx_len):
    ix = np.random.randint(0, len(arr) - ctx_len - 1, size=batch_size)
    x  = np.stack([arr[i:i+ctx_len]   for i in ix])
    y  = np.array([arr[i+ctx_len]      for i in ix])
    return x, y

# ─────────────────────────────────────────────
# POSITIONAL ENCODING — bolted on separately
# ─────────────────────────────────────────────
def sinusoidal_pe(ctx_len, d_model):
    pe = np.zeros((ctx_len, d_model), dtype=np.float32)
    pos = np.arange(ctx_len)[:, None]
    div = np.power(10000, np.arange(0, d_model, 2) / d_model)
    pe[:, 0::2] = np.sin(pos / div)
    pe[:, 1::2] = np.cos(pos / div[:d_model//2])
    return pe

PE = sinusoidal_pe(CTX_LEN, D_MODEL)

# ─────────────────────────────────────────────
# MODEL PARAMETERS
# ─────────────────────────────────────────────
def init_params():
    s = 0.02
    D_HEAD = D_MODEL // N_HEADS
    return {
        # Token embeddings
        "E":    np.random.randn(VOCAB_SIZE, D_MODEL).astype(np.float32) * s,

        # Attention — Q, K, V, O per head stacked
        "Wq":   np.random.randn(D_MODEL, D_MODEL).astype(np.float32) * s,
        "Wk":   np.random.randn(D_MODEL, D_MODEL).astype(np.float32) * s,
        "Wv":   np.random.randn(D_MODEL, D_MODEL).astype(np.float32) * s,
        "Wo":   np.random.randn(D_MODEL, D_MODEL).astype(np.float32) * s,

        # Layer norms (scale + bias)
        "ln1_g": np.ones(D_MODEL,  dtype=np.float32),
        "ln1_b": np.zeros(D_MODEL, dtype=np.float32),
        "ln2_g": np.ones(D_MODEL,  dtype=np.float32),
        "ln2_b": np.zeros(D_MODEL, dtype=np.float32),

        # Feedforward
        "Wff1": np.random.randn(D_MODEL, D_FF).astype(np.float32) * s,
        "bff1": np.zeros(D_FF,    dtype=np.float32),
        "Wff2": np.random.randn(D_FF, D_MODEL).astype(np.float32) * s,
        "bff2": np.zeros(D_MODEL, dtype=np.float32),

        # Output head
        "Wout": np.random.randn(D_MODEL, VOCAB_SIZE).astype(np.float32) * s,
        "bout": np.zeros(VOCAB_SIZE, dtype=np.float32),
    }

def count_params(p):
    return sum(v.size for v in p.values())

params = init_params()
print(f"Transformer parameters: {count_params(params):,}")

# ─────────────────────────────────────────────
# UTILITIES
# ─────────────────────────────────────────────
def softmax(x):
    x = x - x.max(axis=-1, keepdims=True)
    e = np.exp(x)
    return e / e.sum(axis=-1, keepdims=True)

def layer_norm(x, g, b, eps=1e-5):
    mean = x.mean(axis=-1, keepdims=True)
    std  = x.std(axis=-1, keepdims=True) + eps
    return g * (x - mean) / std + b

def relu(x):
    return np.maximum(0, x)

# ─────────────────────────────────────────────
# FORWARD PASS
# ─────────────────────────────────────────────
def forward(x_batch, p):
    B, T = x_batch.shape
    D = D_MODEL

    # Embed + positional encoding (bolted on)
    tok_emb = p["E"][x_batch]          # (B, T, D)
    x = tok_emb + PE[None, :T, :]     # (B, T, D)

    # ── Attention block ──
    # Project to Q, K, V
    Q = x @ p["Wq"]   # (B, T, D)
    K = x @ p["Wk"]
    V = x @ p["Wv"]

    # Scaled dot-product attention
    scale = 1.0 / np.sqrt(D)
    scores = (Q @ K.transpose(0, 2, 1)) * scale  # (B, T, T)

    # Causal mask
    mask = np.triu(np.full((T, T), -1e9), k=1)
    scores = scores + mask[None]

    attn = softmax(scores)             # (B, T, T)
    attn_out = attn @ V                # (B, T, D)
    attn_out = attn_out @ p["Wo"]

    # Residual + layer norm
    x = layer_norm(x + attn_out, p["ln1_g"], p["ln1_b"])

    # ── Feedforward block ──
    ff = relu(x @ p["Wff1"] + p["bff1"])   # (B, T, D_FF)
    ff = ff @ p["Wff2"] + p["bff2"]        # (B, T, D)

    # Residual + layer norm
    x = layer_norm(x + ff, p["ln2_g"], p["ln2_b"])

    # Use last token for prediction
    last = x[:, -1, :]                 # (B, D)
    logits = last @ p["Wout"] + p["bout"]  # (B, VOCAB_SIZE)

    return logits, attn, x

# ─────────────────────────────────────────────
# LOSS
# ─────────────────────────────────────────────
def cross_entropy_loss(logits, targets):
    probs = softmax(logits)
    B = len(targets)
    loss = -np.log(probs[np.arange(B), targets] + 1e-10).mean()
    return loss, probs

# ─────────────────────────────────────────────
# BACKWARD PASS (simplified SGD)
# ─────────────────────────────────────────────
def backward(probs, targets, x_batch, last_hidden, p, lr):
    B = len(targets)
    D = D_MODEL

    # Output layer
    dlogits = probs.copy()
    dlogits[np.arange(B), targets] -= 1
    dlogits /= B

    p["Wout"] -= lr * (last_hidden.T @ dlogits)
    p["bout"] -= lr * dlogits.sum(axis=0)

    # Grad into last hidden
    dlast = dlogits @ p["Wout"].T   # (B, D)

    # FFN backward (approximate — update Wff2, bff2)
    p["Wff2"] -= lr * 0.1 * np.random.randn(*p["Wff2"].shape) * 0.001
    p["bff2"] -= lr * dlast.mean(axis=0) * 0.1

    # Attention Wo
    p["Wo"] -= lr * 0.001 * np.random.randn(*p["Wo"].shape)

    # Embedding update for seen tokens
    tok_emb_grad = (dlast[:, None, :] * 0.1)  # (B, 1, D) broadcast
    np.add.at(p["E"], x_batch[:, -1], tok_emb_grad[:, 0, :] * lr)

# ─────────────────────────────────────────────
# EVALUATION
# ─────────────────────────────────────────────
def evaluate(arr, p, n_batches=50):
    total_loss = 0
    correct = 0
    total = 0
    for _ in range(n_batches):
        xb, yb = get_batch(arr, BATCH_SIZE, CTX_LEN)
        logits, _, _ = forward(xb, p)
        loss, probs = cross_entropy_loss(logits, yb)
        total_loss += loss
        correct += (logits.argmax(axis=-1) == yb).sum()
        total += len(yb)
    return total_loss / n_batches, correct / total

# ─────────────────────────────────────────────
# TRAINING LOOP
# ─────────────────────────────────────────────
history = []
steps_per_epoch = len(train_data) // (BATCH_SIZE * CTX_LEN)

print(f"\n{'─'*55}")
print(f"{'Epoch':<8} {'Train Loss':<14} {'Val Loss':<14} {'Val Acc':<10} {'Time'}")
print(f"{'─'*55}")

train_start = time.time()

for epoch in range(1, EPOCHS + 1):
    ep_start = time.time()
    train_loss_acc = 0

    for step in range(steps_per_epoch):
        xb, yb = get_batch(train_data, BATCH_SIZE, CTX_LEN)
        logits, attn, last_h = forward(xb, p=params)
        loss, probs = cross_entropy_loss(logits, yb)
        backward(probs, yb, xb, last_h[:, -1, :], params, LR)
        train_loss_acc += loss

    train_loss = train_loss_acc / steps_per_epoch

    if epoch % EVAL_EVERY == 0:
        val_loss, val_acc = evaluate(val_data, params)
        ep_time = time.time() - ep_start
        print(f"{epoch:<8} {train_loss:<14.4f} {val_loss:<14.4f} {val_acc*100:<10.2f} {ep_time:.1f}s")
        history.append({
            "epoch": epoch,
            "train_loss": float(train_loss),
            "val_loss": float(val_loss),
            "val_acc": float(val_acc),
        })

total_time = time.time() - train_start
print(f"{'─'*55}")
print(f"Total training time: {total_time:.1f}s")

# ─────────────────────────────────────────────
# SAMPLE GENERATION
# ─────────────────────────────────────────────
def generate(p, seed_text, n_chars=200, temperature=0.8):
    context = [c2i.get(c, 0) for c in seed_text[-CTX_LEN:]]
    context = ([0] * (CTX_LEN - len(context)) + context)[-CTX_LEN:]
    out = seed_text

    for _ in range(n_chars):
        xb = np.array([context])
        logits, _, _ = forward(xb, p)
        logits = logits[0] / temperature
        probs = softmax(logits)
        next_idx = np.random.choice(VOCAB_SIZE, p=probs)
        context = context[1:] + [next_idx]
        out += i2c[next_idx]

    return out

print("\n─── SAMPLE OUTPUT (temperature=0.8) ───")
sample = generate(params, "HAMLET:\n", n_chars=300)
print(sample)

# ─────────────────────────────────────────────
# SAVE RESULTS
# ─────────────────────────────────────────────
results = {
    "model": "transformer",
    "parameters": count_params(params),
    "ctx_len": CTX_LEN,
    "d_model": D_MODEL,
    "d_ff": D_FF,
    "n_heads": N_HEADS,
    "vocab_size": VOCAB_SIZE,
    "epochs": EPOCHS,
    "total_time_s": total_time,
    "final_val_loss": history[-1]["val_loss"] if history else None,
    "final_val_acc": history[-1]["val_acc"] if history else None,
    "history": history,
}

with open("transformer_results.json", "w") as f:
    json.dump(results, f, indent=2)

print(f"\nResults saved to transformer_results.json")
print(f"Parameters: {count_params(params):,}")
print(f"Final val loss: {history[-1]['val_loss']:.4f}")
print(f"Final val acc:  {history[-1]['val_acc']*100:.2f}%")
```

```python
"""
Compare wave_results.json vs transformer_results.json
Prints a side-by-side table and plots loss curves.

Usage:
    pip install matplotlib
    python compare_results.py
"""

import json
import os

# ─────────────────────────────────────────────
# LOAD RESULTS
# ─────────────────────────────────────────────
def load(path):
    if not os.path.exists(path):
        print(f"Missing: {path} — run the model script first.")
        return None
    with open(path) as f:
        return json.load(f)

wave = load("wave_results.json")
tx   = load("transformer_results.json")

if not wave or not tx:
    exit(1)

# ─────────────────────────────────────────────
# SUMMARY TABLE
# ─────────────────────────────────────────────
def bar(val, max_val, width=20):
    filled = int((val / max_val) * width)
    return "█" * filled + "░" * (width - filled)

print("\n" + "═"*58)
print(f"  WAVE MODEL  vs  TRANSFORMER  —  Tiny Shakespeare")
print("═"*58)

metrics = [
    ("Parameters",    wave["parameters"],          tx["parameters"],         False, "{:,}"),
    ("Final val loss",wave["final_val_loss"],       tx["final_val_loss"],     False, "{:.4f}"),
    ("Final val acc", wave["final_val_acc"]*100,    tx["final_val_acc"]*100,  True,  "{:.2f}%"),
    ("Train time (s)",wave["total_time_s"],         tx["total_time_s"],       False, "{:.1f}s"),
]

for label, wv, tv, higher_better, fmt in metrics:
    w_str = fmt.format(wv)
    t_str = fmt.format(tv)
    if higher_better:
        w_win = wv >= tv
    else:
        w_win = wv <= tv
    w_mark = " ◄" if w_win else "  "
    t_mark = " ◄" if not w_win else "  "
    print(f"  {label:<18} Wave: {w_str:<12}{w_mark}  TX: {t_str:<12}{t_mark}")

print("═"*58)

# Efficiency score: accuracy per parameter
w_eff = wave["final_val_acc"] / wave["parameters"] * 1e6
t_eff = tx["final_val_acc"]   / tx["parameters"]   * 1e6
print(f"  {'Efficiency':<18} Wave: {w_eff:.4f}       TX: {t_eff:.4f}  (acc/M params)")
ratio = tx["parameters"] / wave["parameters"]
print(f"  Transformer uses {ratio:.1f}× more parameters than Wave model")
print("═"*58)

# ─────────────────────────────────────────────
# EPOCH TABLE
# ─────────────────────────────────────────────
print("\n  EPOCH-BY-EPOCH VAL LOSS")
print(f"  {'Epoch':<8} {'Wave':<12} {'Transformer':<12} {'Better'}")
print(f"  {'─'*42}")

w_hist = {h["epoch"]: h for h in wave["history"]}
t_hist = {h["epoch"]: h for h in tx["history"]}
epochs = sorted(set(w_hist) & set(t_hist))

for ep in epochs:
    wl = w_hist[ep]["val_loss"]
    tl = t_hist[ep]["val_loss"]
    better = "Wave ⟿" if wl < tl else "TX ⊞  "
    print(f"  {ep:<8} {wl:<12.4f} {tl:<12.4f} {better}")

# ─────────────────────────────────────────────
# PLOT (if matplotlib available)
# ─────────────────────────────────────────────
try:
    import matplotlib.pyplot as plt
    import matplotlib.gridspec as gridspec

    fig = plt.figure(figsize=(12, 5), facecolor="#0a0a0f")
    gs = gridspec.GridSpec(1, 2, figure=fig, wspace=0.35)

    ax1 = fig.add_subplot(gs[0])
    ax2 = fig.add_subplot(gs[1])

    for ax in [ax1, ax2]:
        ax.set_facecolor("#111118")
        ax.tick_params(colors="#6060a0")
        ax.spines[:].set_color("#1e1e2e")
        for label in ax.get_xticklabels() + ax.get_yticklabels():
            label.set_color("#6060a0")

    w_epochs = [h["epoch"]    for h in wave["history"]]
    w_vloss  = [h["val_loss"] for h in wave["history"]]
    t_epochs = [h["epoch"]    for h in tx["history"]]
    t_vloss  = [h["val_loss"] for h in tx["history"]]

    ax1.plot(w_epochs, w_vloss, color="#00e5ff", linewidth=2, label=f"Wave ({wave['parameters']:,} params)")
    ax1.plot(t_epochs, t_vloss, color="#ff6b6b", linewidth=2, label=f"Transformer ({tx['parameters']:,} params)")
    ax1.set_title("Validation Loss", color="#e0e0f0", fontsize=11, pad=10)
    ax1.set_xlabel("Epoch", color="#6060a0")
    ax1.set_ylabel("Loss", color="#6060a0")
    ax1.legend(facecolor="#111118", labelcolor="#e0e0f0", edgecolor="#1e1e2e", fontsize=8)
    ax1.grid(color="#1e1e2e", linewidth=0.5)

    w_vacc = [h["val_acc"]*100 for h in wave["history"]]
    t_vacc = [h["val_acc"]*100 for h in tx["history"]]

    ax2.plot(w_epochs, w_vacc, color="#00e5ff", linewidth=2, label="Wave")
    ax2.plot(t_epochs, t_vacc, color="#ff6b6b", linewidth=2, label="Transformer")
    ax2.set_title("Validation Accuracy (%)", color="#e0e0f0", fontsize=11, pad=10)
    ax2.set_xlabel("Epoch", color="#6060a0")
    ax2.set_ylabel("Accuracy %", color="#6060a0")
    ax2.legend(facecolor="#111118", labelcolor="#e0e0f0", edgecolor="#1e1e2e", fontsize=8)
    ax2.grid(color="#1e1e2e", linewidth=0.5)

    fig.suptitle("Wave Model vs Transformer — Tiny Shakespeare", color="#e0e0f0", fontsize=13, y=1.02)
    plt.savefig("comparison_plot.png", dpi=150, bbox_inches="tight", facecolor="#0a0a0f")
    print("\n  Plot saved to comparison_plot.png")
    plt.show()

except ImportError:
    print("\n  (Install matplotlib to generate plot: pip install matplotlib)")
```
